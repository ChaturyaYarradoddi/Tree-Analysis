```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.metrics import classification_report, accuracy_score, roc_auc_score
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import f1_score
from sklearn.tree import plot_tree
from sklearn.metrics import  ConfusionMatrixDisplay

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from dtreeviz import model
from cairosvg import svg2png
from IPython.display import Image, display

from statsmodels.stats.outliers_influence import variance_inflation_factor

pd.set_option("display.max_columns", 110)
pd.set_option("display.max_rows", 25970)
```

```python
# Read an Excel file into a DataFrame using pandas

df = pd.read_excel("/content/drive/MyDrive/UMBCDATA/USM FY 2014 New FT and Transfers Retention Degree and FA  12-21-23 1.xlsx")
```

```python
# Display the first 5 rows of the DataFrame to get a quick look at the data
df.head()
```

```python
# Returns (rows, columns) of the DataFrame
df.shape
```

```python
# Check percentage of missing values in each column
null_percentage = (df.isnull().sum() / len(df)) * 100

# Keep only columns with missing values and sort by percentage
null_percentage = null_percentage[null_percentage > 0].sort_values(ascending=False)

# Display the missing data report
print("Columns with missing values (sorted):")
print(null_percentage)
```

```python
#  Fill financial aid columns (assume missing aid = 0)
grant_cols = [col for col in df.columns if 'GrantAmountFY' in col]
merit_cols = [col for col in df.columns if 'MeritAmountFY' in col]
df[grant_cols] = df[grant_cols].fillna(0)
df[merit_cols] = df[merit_cols].fillna(0)
```

```python
# Columns related to graduation logic
grad_date_cols = [f'degreedate{i}' for i in range(1, 5)]
grad_flag_cols = [f'deg{i}' for i in range(1, 5)]
hegis_cols = ['HegisDeg1'] + [f'hegisdeg{i}' for i in range(2, 5)]
fy_cols = [f'FY{i}' for i in range(1, 5)]

# Fill degree dates with 0 (no graduation date)
df[grad_date_cols] = df[grad_date_cols].fillna(0)

# Fill degree flags and hegis codes with 0 (no degree)
df[grad_flag_cols + hegis_cols + fy_cols] = df[grad_flag_cols + hegis_cols + fy_cols].fillna(0)

# Confirm all nulls handled
print("Remaining nulls in graduation-related columns:")
print(df[grad_date_cols + grad_flag_cols + hegis_cols + fy_cols].isnull().sum())
```

```python
#  Fill major columns ('0' for no major declared)
major_cols = [col for col in df.columns if 'Major' in col and 'FY' in col]
df[major_cols] = df[major_cols].fillna(0)
```

```python
# Fill other demographic fields
for col in ['EFCAIDFY', 'CommuteStatus']:
    if col in df.columns:
        df[col] = df[col].fillna("Unknown")
```

```python
# Define list of HEGIS prefixes for STEM majors
stem_prefixes = ('7', '9', '4', '17', '19')

# Function to check if EntryMajor starts with any STEM prefix
df['Stem1'] = df['EntryMajor'].astype(str).apply(
    lambda x: 'Yes' if x.startswith(stem_prefixes) else 'No'
)

# Quick summary
print("STEM Major Distribution:")
print(df['Stem1'].value_counts())
```

```python
#  Graduation status
def check_graduation(row):
    for i in range(1, 5):
        if pd.notnull(row[f'degreedate{i}']) and row[f'degreedate{i}'] != 0:
            return 'Yes'
    return 'No'
df['Graduated'] = df.apply(check_graduation, axis=1)
```

```python
#  Total support and flags
df['totalsupport'] = df[grant_cols + merit_cols].sum(axis=1)
df['supportedY1'] = df.apply(lambda x: 1 if (x['GrantAmountFY2014'] + x['MeritAmountFY2014']) > 0 else 0, axis=1)
df['supported'] = df['totalsupport'].apply(lambda x: 1 if x > 0 else 0)
```

```python
#  Categorize EFC
def categorize_efc(val):
    if val == -1:
        return 'Unknown'
    elif val < 6657:
        return 'High'
    elif 6657 <= val <= 13312:
        return 'Medium'
    else:
        return 'Low'
df['EFC_Category'] = df['EFC'].apply(categorize_efc)
```

```python
#  Determine "In Need" status
need_cols = [col for col in df.columns if 'Need' in col and 'FY' in col]
need_values = ['High', 'Medium', 'Low']
df['In_Need'] = df[need_cols].apply(lambda row: 'Yes' if any(val in need_values for val in row.values) else 'No', axis=1)
```

```python
#  Full-time enrollment flag
sem_cols = [col for col in df.columns if col.startswith('FY') and ('Fall' in col or 'Spring' in col)]
df['Is_FullTime'] = df[sem_cols].apply(lambda row: 'Yes' if 'FT' in row.values else 'No', axis=1)
```

```python
# Function to get the first declared major (ignoring nulls and 0s)
def get_first_major(row):
    majors = [m for m in row if pd.notnull(m) and m != 0]
    return majors[0] if majors else None  # Return first valid major

# Function to get the last declared major (ignoring nulls and 0s)
def get_last_major(row):
    majors = [m for m in row if pd.notnull(m) and m != 0]
    return majors[-1] if majors else None  # Return last valid major

# Check if the student switched majors in the middle but ended on the same major
def changed_in_middle(row):
    majors = [m for m in row if pd.notnull(m) and m != 0]
    return 'Yes' if len(set(majors)) > 1 and majors[0] == majors[-1] else 'No'

# Count the number of times a student switched majors
def count_major_switches(row):
    majors = [m for m in row if pd.notnull(m) and m != 0]
    return sum(majors[i] != majors[i+1] for i in range(len(majors)-1))

# Apply the above functions to compute major change-related features
df['Initial_Major'] = df[major_cols].apply(get_first_major, axis=1)  # First valid major
df['Final_Major'] = df[major_cols].apply(get_last_major, axis=1)     # Last valid major

# Binary column: Did the student change their major from start to end?
df['Changed_Major'] = df.apply(
    lambda x: 'Yes' if x['Initial_Major'] and x['Final_Major'] and x['Initial_Major'] != x['Final_Major'] else 'No',
    axis=1
)

# Did the student change majors in the middle but start and end with the same one?
df['Changed_In_Middle'] = df[major_cols].apply(changed_in_middle, axis=1)

# Total number of unique majors over the years
df['Num_Majors'] = df[major_cols].apply(
    lambda row: len(set([m for m in row if pd.notnull(m) and m != 0])),
    axis=1
)

# Number of times the student changed majors (based on year-to-year differences)
df['Major_Change_Count'] = df[major_cols].apply(count_major_switches, axis=1)
```

```python
# Got_Grant: Mark 'Yes' if the student received any grant amount in any year; otherwise 'No'
df['Got_Grant'] = df[grant_cols].sum(axis=1).apply(lambda x: 'Yes' if x > 0 else 'No')

# Got_Merit: Mark 'Yes' if the student received any merit scholarship in any year; otherwise 'No'
df['Got_Merit'] = df[merit_cols].sum(axis=1).apply(lambda x: 'Yes' if x > 0 else 'No')
```

```python
# Preview result
df[['Graduated', 'Initial_Major', 'Final_Major', 'Changed_Major', 'Got_Grant', 'Got_Merit', 'supportedY1', 'In_Need', 'Is_FullTime','totalsupport']].head()
```

```python
# Select features and target
target = 'Graduated'
features = ['totalsupport', 'supportedY1', 'supported', 'EFC_Category', 'In_Need', 'Is_FullTime', 'Changed_Major', 'Changed_In_Middle', 'Got_Grant', 'Got_Merit']
X = df[features]
y = df[target].map({'Yes': 1, 'No': 0})
```

```python
# Handle categorical variables
cat_cols = ['EFC_Category', 'In_Need', 'Is_FullTime', 'Changed_Major', 'Changed_In_Middle', 'Got_Grant', 'Got_Merit']
num_cols = ['totalsupport', 'supportedY1', 'supported']
```

```python
# Raw XGBoost
xgb_raw_preprocessor = ColumnTransformer([
    ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), cat_cols)
], remainder='passthrough')

xgb_raw = Pipeline([
    ('preprocessor', xgb_raw_preprocessor),
    ('classifier', XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42))
])

X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, test_size=0.2, random_state=42)
xgb_raw.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay, precision_score, recall_score, f1_score, roc_auc_score
```

```python
# Raw XGBoost Metrics and Feature Importance
y_pred_raw = xgb_raw.predict(X_test)
print("[Raw XGBoost] Accuracy:", accuracy_score(y_test, y_pred_raw))
print("[Raw XGBoost] Precision:", precision_score(y_test, y_pred_raw))
print("[Raw XGBoost] Recall:", recall_score(y_test, y_pred_raw))
print("[Raw XGBoost] F1 Score:", f1_score(y_test, y_pred_raw))
print("[Raw XGBoost] ROC AUC:", roc_auc_score(y_test, y_pred_raw))
ConfusionMatrixDisplay.from_predictions(y_test, y_pred_raw, display_labels=['Not Graduated', 'Graduated'], cmap='Blues')
plt.title("Confusion Matrix - Raw XGBoost")
plt.show()
```

```python
feature_names_raw = [name.split('__')[-1] for name in xgb_raw.named_steps['preprocessor'].get_feature_names_out()]
importances_raw = xgb_raw.named_steps['classifier'].feature_importances_
plt.figure(figsize=(10, 6))
plt.barh(feature_names_raw, importances_raw)
plt.title("XGBoost Feature Importance - Raw")
plt.tight_layout()
plt.savefig("feature_importance_xgboost_raw.png")
plt.show()
```

```python

# Scaled XGBoost
xgb_scaled_preprocessor = ColumnTransformer([
    ('num', StandardScaler(), num_cols),
    ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), cat_cols)
])

xgb_scaled = Pipeline([
    ('preprocessor', xgb_scaled_preprocessor),
    ('classifier', XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42))
])
xgb_scaled.fit(X_train, y_train)
```

```python
# Scaled XGBoost Metrics and Feature Importance
y_pred_scaled = xgb_scaled.predict(X_test)
print("[Scaled XGBoost] Accuracy:", accuracy_score(y_test, y_pred_scaled))
print("[Scaled XGBoost] Precision:", precision_score(y_test, y_pred_scaled))
print("[Scaled XGBoost] Recall:", recall_score(y_test, y_pred_scaled))
print("[Scaled XGBoost] F1 Score:", f1_score(y_test, y_pred_scaled))
print("[Scaled XGBoost] ROC AUC:", roc_auc_score(y_test, y_pred_scaled))
ConfusionMatrixDisplay.from_predictions(y_test, y_pred_scaled, display_labels=['Not Graduated', 'Graduated'], cmap='Greens')
plt.title("Confusion Matrix - Scaled XGBoost")
plt.show()
```

```python
feature_names_scaled = [name.split('__')[-1] for name in xgb_scaled.named_steps['preprocessor'].get_feature_names_out()]
importances_scaled = xgb_scaled.named_steps['classifier'].feature_importances_
plt.figure(figsize=(10, 6))
plt.barh(feature_names_scaled, importances_scaled)
plt.title("XGBoost Feature Importance - Scaled")
plt.tight_layout()
plt.savefig("feature_importance_xgboost_scaled.png")
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier
from dtreeviz import model
from cairosvg import svg2png
import pandas as pd
from IPython.display import Image, display

def xgboost_to_dtreeviz(xgb_model, X_train, y_train, preprocessor, feature_names, label='scaled', dpi=300):
    """
    Approximates an XGBoost model using a DecisionTreeClassifier and visualizes it using dtreeviz.
    This is helpful for interpreting model logic in a simplified form.

    Parameters:
    - xgb_model: trained XGBClassifier inside a pipeline
    - X_train: raw training feature DataFrame
    - y_train: training target labels
    - preprocessor: fitted ColumnTransformer used in training
    - feature_names: list of feature names after transformation
    - label: string label used for saving output files
    - dpi: resolution for PNG export
    """

    # Apply preprocessing (scaling + encoding)
    X_trans = preprocessor.fit_transform(X_train)

    # Convert sparse matrix to dense if needed
    if hasattr(X_trans, "toarray"):
        X_trans = X_trans.toarray()

    # Convert to DataFrame with feature names for interpretability
    X_df = pd.DataFrame(X_trans, columns=feature_names)

    # Train a small decision tree to approximate XGBoost behavior
    approx_tree = DecisionTreeClassifier(max_depth=4, random_state=42)
    approx_tree.fit(X_df, y_train)

    # Generate dtreeviz object
    viz = model(
        approx_tree,
        X_df,
        y_train,
        feature_names=feature_names,
        target_name="Graduated",
        class_names={0: "Not Graduated", 1: "Graduated"}
    )

    # Render visualization and export as SVG
    render = viz.view()
    svg = render.svg()

    # Define paths
    svg_path = f"dtreeviz_xgboost_{label}.svg"
    png_path = f"dtreeviz_xgboost_{label}.png"

    # Save SVG
    with open(svg_path, "w", encoding="utf-8") as f:
        f.write(svg)

    # Convert SVG to PNG
    svg2png(bytestring=svg.encode("utf-8"), write_to=png_path, dpi=dpi)

    # Display result
    print(f"DTreeViz approximation of XGBoost ({label}) saved to {png_path}")
    display(Image(filename=png_path))
```

```python
# Extract cleaned-up feature names
feature_names_raw = [name.split('__')[-1] for name in xgb_raw.named_steps['preprocessor'].get_feature_names_out()]

# Visualize the simplified XGBoost tree using dtreeviz on raw features
xgboost_to_dtreeviz(
    xgb_model=xgb_raw.named_steps['classifier'],  # Optional but consistent
    X_train=X_train,                              # Raw training features
    y_train=y_train,                              # Raw labels
    preprocessor=xgb_raw.named_steps['preprocessor'],  # The preprocessor from the pipeline
    feature_names=feature_names_raw,              # Clean feature names
    label='raw'                                   # For saving files like dtreeviz_xgboost_raw.png
)
```

```python
# Apply to scaled
feature_names_scaled = [name.split('__')[-1] for name in xgb_scaled.named_steps['preprocessor'].get_feature_names_out()]
xgboost_to_dtreeviz(xgb_model=xgb_scaled.named_steps['classifier'], X_train=X_train, y_train=y_train, preprocessor=xgb_scaled.named_steps['preprocessor'], feature_names=feature_names_scaled, label='scaled')
```

```python
from sklearn.tree import DecisionTreeClassifier, export_text
import pandas as pd

def print_human_readable_tree_raw(xgb_pipeline, X_train, y_train, feature_names, max_depth=4, label='raw'):
    """
    Fits a DecisionTreeClassifier to approximate the XGBoost model behavior using raw input,
    and prints a human-readable textual version of the tree.

    Parameters:
    - xgb_pipeline: Trained XGBoost pipeline (should include preprocessor + classifier)
    - X_train: Raw training features (DataFrame)
    - y_train: Labels
    - feature_names: Cleaned feature names after transformation
    - max_depth: Max depth of the approximated tree
    - label: Identifier label for printing
    """
    # Extract preprocessor
    preprocessor = xgb_pipeline.named_steps['preprocessor']

    # Transform X_train
    X_trans = preprocessor.transform(X_train)
    if hasattr(X_trans, "toarray"):
        X_trans = X_trans.toarray()

    # Build DataFrame with feature names
    X_df = pd.DataFrame(X_trans, columns=feature_names)

    # Fit simplified Decision Tree
    tree = DecisionTreeClassifier(max_depth=max_depth, random_state=42)
    tree.fit(X_df, y_train)

    # Print readable tree
    tree_text = export_text(tree, feature_names=feature_names)
    print(f"\n[Human Readable Decision Tree Approximation ({label})]:\n")
    print(tree_text)
```

```python
print_human_readable_tree_raw(
    xgb_pipeline=xgb_raw,
    X_train=X_train,
    y_train=y_train,
    feature_names=feature_names_raw,
    max_depth=4,
    label='raw'
)
```
